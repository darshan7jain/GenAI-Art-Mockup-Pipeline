```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://www.memeraki.com/collections/mata-ni-pachedi?sort_by=best-selling'

page = requests.get(url)

soup = BeautifulSoup(page.text, 'html')
```

```python
lnk=soup.find('ul',id='id="c-accordion--1603134-1"')
```

```python
ol_data=soup.find('ol',id='main-collection-product-grid')
```

```python
li_items = ol_data.find_all('li')[:10]
```

```python
#Single artwork

for item in li_items:
    product_name = item.get('data-alpha')
    size = "N/A"
    details_div = item.find('div', style=lambda val: val and 'justify-content: space-between' in val)
    if details_div:
        p_tags = details_div.find_all('p')
        if len(p_tags) >= 2:
            size = p_tags[1].get_text(strip=True)

    price_tag = item.find('dl', class_='price__regular')
    price = price_tag.get_text(strip=True) if price_tag else "N/A"

    print(f"Product Name: {product_name}")
    print(f"Size: {size}")
    print(f"Price: {price}")
    print("---")
```

```python
artnames=['madhubani','kalamkari','pichwai','gond-art','kalighat','bengal-pattachitra','miniature-paintings','mata-ni-pachedi','warli-paintings','blue-pottery','phad','kerala-mural','tanjore','lippan-kaam']
```

```python
#Final list of all artworks
art_data_dict = {}

for artname in artnames:
    url = f'https://www.memeraki.com/collections/{artname}?sort_by=best-selling'
    page = requests.get(url)
    soup = BeautifulSoup(page.text, 'html')
    
    ol_data = soup.find('ol', id='main-collection-product-grid')
    if not ol_data:
        art_data_dict[artname] = []
        continue
    
    li_items = ol_data.find_all('li')[:10]
    data = []
    
    for item in li_items:
        product_name = item.get('data-alpha')
        size = "N/A"
        details_div = item.find('div', style=lambda val: val and 'justify-content: space-between' in val)
        if details_div:
            p_tags = details_div.find_all('p')
            if len(p_tags) >= 2:
                size = p_tags[1].get_text(strip=True)
        price_tag = item.find('dl', class_='price__regular')
        price = price_tag.get_text(strip=True) if price_tag else "N/A"
        data.append({
            'Product Name': product_name,
            'Size': size,
            'Price': price
        })
    art_data_dict[artname] = data

print(art_data_dict)
# import pprint
# pprint.pprint(art_data_dict)
```
